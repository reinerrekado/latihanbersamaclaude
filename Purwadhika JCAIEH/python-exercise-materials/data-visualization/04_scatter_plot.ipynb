# Scatter Plot

A scatter plot places one point per observation using two numeric
variables as its x/y coordinates. It's the go-to chart for spotting a
**relationship** (correlation) or **pattern** between two continuous
variables, and for eyeballing outliers or clusters.

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
tips.head(3)
```

## Matplotlib

```python
plt.figure(figsize=(8, 6))
plt.scatter(tips['total_bill'], tips['tip'])
plt.title('Scatter Plot: Total Bill vs Tip', size=15)
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

## Seaborn

`hue=` and `style=` layer a third and fourth variable onto the same 2D plot using color and marker shape.

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='day', style='time')
plt.title('Scatter Plot: Total Bill vs Tip', size=15)
plt.show()
```

## Pandas

```python
tips.plot.scatter(x='total_bill', y='tip')
plt.title('Scatter Plot: Total Bill vs Tip', size=15)
plt.show()
```

## When to avoid

- When your data isn't bi-dimensional (need at least two numeric
  variables).
- When you actually care about a trend **over time** - use a line
  plot instead.
- When one of your variables is categorical (a handful of groups) -
  the points just stack into vertical strips and don't reveal much.
