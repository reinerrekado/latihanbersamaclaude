# Bar Plot

A bar chart represents each category with a rectangular bar, whose
length is proportional to its value. Bars can run vertically (a
"column chart") or horizontally. It's the most common chart for
**comparing discrete categories**.

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
tips_by_day = tips[['tip', 'day']].groupby(['day']).mean()
tips_by_day
```

## Matplotlib

```python
x = tips['day'].unique().tolist()
y = tips_by_day['tip']

plt.figure(figsize=(8, 6))
plt.bar(x, y, color=['green', 'blue', 'red', 'yellow'])
plt.title('Average Tip by Day', size=15)
plt.xlabel('Day')
plt.ylabel('Tip')
plt.show()

# plt.barh(...) draws the same chart horizontally
```

## Seaborn

`hue=` groups bars side by side per category.

```python
sns.barplot(data=tips, x='day', y='tip', hue='sex')
plt.title('Average Tip by Day and Gender', size=15)
plt.legend(loc=0)
plt.show()
```

## Pandas

```python
tips_by_day.plot.bar()
plt.title('Average Tip by Day', size=15)
plt.show()
```

## When to avoid

Simple bar charts struggle when several categories together make up a
whole (a pie chart or 100% stacked bar communicates "share of total"
better), or when you have many time periods to compare (a line plot
shows a trend more clearly than a wall of bars).
