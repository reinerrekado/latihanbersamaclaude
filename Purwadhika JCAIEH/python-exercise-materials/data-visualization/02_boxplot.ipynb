# Boxplot (Box-and-Whisker Plot)

A boxplot summarizes a numeric variable with its **five-number
summary**: minimum, first quartile (Q1), median, third quartile (Q3),
and maximum - and flags likely outliers.

- **Box**: spans Q1 to Q3 (the middle 50% of the data), with a line at
  the median.
- **IQR** (interquartile range) = Q3 - Q1.
- **Whiskers**: extend to the smallest/largest values that are still
  within `1.5 * IQR` of the box.
- **Outliers**: individual points plotted beyond the whiskers - i.e.
  below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`.

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
tips.head(3)
```

## Matplotlib

```python
plt.figure(figsize=(8, 6))
plt.boxplot(tips['total_bill'])
plt.title('Total Bill Boxplot', size=15)
plt.ylabel('Value')
plt.show()
```

## Seaborn

Split the boxes out by a categorical column with `y=`.

```python
sns.boxplot(x=tips['tip'], y=tips['day'])
plt.title('Tip Boxplot by Day', size=20)
plt.show()
```

## Pandas

```python
tips.boxplot(column='tip')
plt.title('Tip Boxplot', size=20)
plt.show()
```

## Reading the shape

- Median centered, whiskers roughly equal -> **symmetric** distribution.
- Median closer to the bottom of the box, shorter lower whisker ->
  **right-skewed** (positively skewed).
- Median closer to the top of the box, shorter upper whisker ->
  **left-skewed** (negatively skewed).
