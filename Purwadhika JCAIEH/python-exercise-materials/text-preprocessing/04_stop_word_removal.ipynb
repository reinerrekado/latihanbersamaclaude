# Stop Word Removal

**Objective:** remove common words that don't carry significant meaning.

**What are stop words?**
- High-frequency words: `"the"`, `"is"`, `"at"`, `"which"`, `"on"`
- Often grammatical rather than semantic
- Language-specific

**Benefits:** smaller vocabulary, focuses on meaningful content words,
faster processing.

**Caution:** may remove important context - e.g. in sentiment analysis,
`"not good"` vs `"good"` (removing `"not"` flips the meaning).

```python
import nltk

nltk.download('stopwords', quiet=True)
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

stop_words = set(stopwords.words('english'))
print("number of English stop words:", len(stop_words))
print("a few examples:", list(stop_words)[:10])
```

```python
text = "This is a sample sentence with stop words"
tokens = word_tokenize(text.lower())

filtered_tokens = [word for word in tokens if word not in stop_words]

print("tokens:         ", tokens)
print("filtered tokens:", filtered_tokens)
```

## Caution in practice: negation gets removed too

```python
text = "This movie is not good"
tokens = word_tokenize(text.lower())
filtered_tokens = [word for word in tokens if word not in stop_words]

print("tokens:         ", tokens)
print("filtered tokens:", filtered_tokens)
# "not" is a stop word by default, so it disappears here - the filtered
# version reads as neutral/positive ("movie good") when the original was
# negative. For sentiment tasks, consider keeping negation words.
```
