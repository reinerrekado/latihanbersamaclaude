# Text Cleaning

**Objective:** remove or standardize unwanted characters and formatting
that don't carry meaning for the task.

**Common operations:**
- **Remove HTML tags:** `<p>Hello</p>` -> `Hello`
- **Remove URLs:** links rarely help a text-classification model
- **Handle special characters:** keep only alphanumeric and essential punctuation
- **Remove extra whitespace:** multiple spaces/tabs/newlines -> a single space

```python
import re

def clean_text(text):
    # Remove HTML tags
    text = re.sub(r'<[^>]+>', '', text)

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # Remove extra whitespace
    text = re.sub(r'\s+', ' ', text).strip()

    return text


sample = "<p>Hi,    my name is   Rama!</p>  Check my site: http://example.com "
print("before:", repr(sample))
print("after: ", repr(clean_text(sample)))
```

## Stripping punctuation / special characters

Depending on the task, you may also want to drop punctuation and
symbols entirely, keeping only letters, digits, and spaces.

```python
def strip_punctuation(text):
    return re.sub(r'[^a-zA-Z0-9\s]', '', text)


tweet = "WOW!!! This is amazing... 10/10 would recommend :) #bestpurchase"
print("before:", tweet)
print("after: ", strip_punctuation(tweet))
```

**Caution:** be careful with tasks like sentiment analysis - stripping
`!` and `?` throws away signal (excitement, emphasis) that can matter.
Always match cleaning aggressiveness to what your task actually needs.
