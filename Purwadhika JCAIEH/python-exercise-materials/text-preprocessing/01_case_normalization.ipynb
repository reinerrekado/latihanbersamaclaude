# Case Normalization

**Objective:** standardize letter case so the same word isn't
treated as several different words just because of capitalization.

**Approaches:**
- **Lowercase** - most common: `"Hello"` -> `"hello"`
- **Uppercase** - less common: `"Hello"` -> `"HELLO"`
- **Title case** - for specific applications: `"hello world"` -> `"Hello World"`

**Considerations:**
- Preserves meaning in most cases
- May lose information (proper nouns, acronyms - `"US"` the country vs `"us"` the pronoun)
- Essential for reducing vocabulary size

```python
text = "Hi, my name is Rama!"

print("original:", text)
print("lower():", text.lower())
print("upper():", text.upper())
print("title():", text.title())
```

## Why it matters: vocabulary size

Without normalization, a model sees `"Great"`, `"great"`, and
`"GREAT"` as three unrelated words. After lowercasing, they collapse
into one.

```python
variants = ["Great", "great", "GREAT", "GReAt"]

print("unique words before lowercasing:", len(set(variants)))
print("unique words after lowercasing: ", len(set(w.lower() for w in variants)))
```
