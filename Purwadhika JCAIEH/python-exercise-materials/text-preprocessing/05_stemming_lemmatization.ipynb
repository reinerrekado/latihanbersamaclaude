# Stemming and Lemmatization

**Objective:** reduce words to their root or base forms.

**Stemming**
- Approach: remove suffixes using rules
- Fast but crude: `"running"` -> `"run"`, `"flies"` -> `"fli"`
- Popular algorithms: Porter Stemmer, Snowball Stemmer

**Lemmatization**
- Approach: use a dictionary and grammatical analysis
- Accurate but slower: `"running"` -> `"run"`, `"better"` -> `"good"`
- Considers context: part of speech matters

```python
import nltk

nltk.download('wordnet', quiet=True)
nltk.download('omw-1.4', quiet=True)

from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

words = ["running", "flies", "dying", "lying", "studying"]

stemmed = [stemmer.stem(word) for word in words]
lemmatized = [lemmatizer.lemmatize(word, pos='v') for word in words]

print("original:  ", words)
print("stemmed:   ", stemmed)
print("lemmatized:", lemmatized)
```

## Lemmatization needs a part of speech to be accurate

`WordNetLemmatizer` defaults to treating a word as a noun unless you
tell it otherwise with `pos=`. This also shows the crude-vs-accurate
gap between the two techniques: the stemmer can't turn `"better"` into
`"good"` because it only strips suffixes, it has no dictionary.

```python
word = "better"

print("stemmer.stem('better'):                 ", stemmer.stem(word))
print("lemmatizer.lemmatize('better'):          ", lemmatizer.lemmatize(word))
print("lemmatizer.lemmatize('better', pos='a'):", lemmatizer.lemmatize(word, pos='a'))
# pos='a' tells the lemmatizer "better" is an adjective here, which is
# what lets it look up "good" as the base form.
```
