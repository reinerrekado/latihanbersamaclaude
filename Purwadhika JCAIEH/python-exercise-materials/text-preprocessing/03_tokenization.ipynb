# Tokenization

**Objective:** split text into individual units (tokens) for processing.

**Types of tokenization:**
- **Word tokenization:** split by spaces and punctuation
- **Sentence tokenization:** split into sentences
- **Subword tokenization:** split into smaller meaningful units (used by
  modern models like BPE/WordPiece), e.g. `"unhappiness"` -> `["un", "happy", "ness"]`

(Install nltk first if you haven't: `pip install nltk`)

````python
import nltk

# Only needs to run once - downloads the tokenizer models nltk uses
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

from nltk.tokenize import word_tokenize, sent_tokenize
````

````python
text = "Hello, world! How are you?"

tokens = word_tokenize(text)
print("word_tokenize:", tokens)
````

````python
paragraph = "Hello world. How are you? I'm fine."

sentences = sent_tokenize(paragraph)
print("sent_tokenize:", sentences)
````

## Why not just `text.split()`?

A naive `.split()` on whitespace leaves punctuation stuck to words -
`"world!"` and `"world"` end up as different tokens. `word_tokenize`
understands punctuation and contractions and separates them out.

````python
text = "Hello, world! How are you?"

print("str.split():   ", text.split())
print("word_tokenize:", word_tokenize(text))
````

## Subword tokenization (conceptual)

Word-level tokenization struggles with rare or made-up words - if
`"unhappiness"` never appeared in training data, the model has no idea
what to do with it. Subword tokenizers split words into smaller,
frequently-seen chunks instead:

```
"unhappiness" -> ["un", "happy", "ness"]
```

This is what algorithms like **Byte-Pair Encoding (BPE)** and
**WordPiece** do under the hood in libraries like `transformers` - out
of scope for this class, but good to know exists.
