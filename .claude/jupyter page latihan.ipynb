Tulis program yang menerima input suhu dalam Fahrenheit, lalu ubah jadi Celsius. 
Rumusnya: (F - 32) * 5/9.

```python
def konversi_ke_celsius(fahrenheit):
    celsius = (fahrenheit - 32) * 5/9
    return celsius

print (konversi_ke_celsius (32))
```

```python
def convertTemperature(Celsius):
    Kelvin = Celsius + 273.15
    Fahrenheit = Celsius * 1.80 + 32.00
    ans = [Kelvin, Fahrenheit]

    return ans

print (convertTemperature (32))
```

```python
class Solution:
    def convertTemperature(self, celsius: float) -> List[float]:
        kelvin = celsius + 273.15
        fahrenheit = celsius * 1.80 + 32.00
        ans = [kelvin, fahrenheit]

        return ans

sol = Solution() 
print(sol.convertTemperature(32)) 
```

Tulis program yang mengonversi centimeter menjadi kilometer. Format output-nya khusus: bukan cuma angka, tapi digabung teks "km".
Contoh: input 100000 (cm) → output harus "1 km" (string, bukan cuma angka 1).

```python
def konversi_ke_kilometer(cm):
    km = (cm / 100000)
    return km

print (konversi_ke_kilometer (100000), "km")
```

Soal 3: Fungsi Cek Ganjil/Genap

Tulis fungsi yang menerima input integer n, dan mengembalikan (return) nilai True kalau n itu ganjil (odd), atau False kalau n itu genap (even).
(Catatan: contoh di soal aslinya sempat nyebut format uang Rupiah seperti "Rp. 1.000,00" — itu kemungkinan typo dari pembuat soal. Fokus aja ke logika boolean ganjil/genap sesuai instruksi utama.)

```python
def fungsi_Mengecek_Ganjil(n):
    if n % 2 == 1:
        return True
    else:
        return False

print(fungsi_Mengecek_Ganjil(5))  # Output: True
```

```python
def fungsi_Mengecek_Genap(n):
    if n % 2 == 0:
        return True
    else:
        return False

print(fungsi_Mengecek_Genap(10))  # Output: True
```
