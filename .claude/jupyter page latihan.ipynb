```python
class Solution:
    def containsDuplicate(self, nums: List[int]) -> bool:
        sudah_dilihat = set()          # "buku catatan" angka yang sudah pernah muncul, mulai kosong

        for angka in nums:               # loop SATU variabel saja — nggak butuh index, jadi nggak perlu enumerate()
            if angka in sudah_dilihat:    # cek: apakah angka INI SENDIRI sudah pernah dicatat sebelumnya?
                return True               # sudah pernah muncul = ketemu kembar, langsung berhenti

            sudah_dilihat.add(angka)      # belum pernah → catat sekarang, biar angka berikutnya bisa ngecek ke sini

        return False                      # loop selesai TOTAL tanpa pernah return True → berarti tidak ada kembar
Solution().containsDuplicate([1,2,3,1])
```
