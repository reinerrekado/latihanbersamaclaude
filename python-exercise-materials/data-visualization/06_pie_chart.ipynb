# Pie Chart

A pie chart is a circle divided into slices, where each slice's size
represents its proportion of a whole. It only makes sense when the
slices actually sum to 100% of something.

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
tips_by_day = tips[['tip', 'day']].groupby(['day'], as_index=False).mean()
tips_by_day
```

## Matplotlib

```python
plt.figure(figsize=(8, 6))
plt.pie(
    tips_by_day['tip'],
    labels=tips_by_day['day'],
    autopct='%.2f',            # show the value inside each slice
    explode=(0, 0.2, 0, 0.1),  # pull specific slices out
)
plt.axis('equal')
plt.title('Average Tip by Day', size=15)
plt.show()
```

## Pandas

```python
tips_by_day.plot.pie(
    y='tip',
    labels=tips_by_day['day'],
    autopct='%.2f',
    figsize=(8, 6),
)
plt.title('Average Tip by Day', size=15)
plt.show()
```

## When to use / avoid

- **Use** it to show how a handful of categories split up one whole
  (e.g. each division's share of total company revenue).
- **Avoid** it when: values change over time (use a line plot
  instead), the slices don't sum to a meaningful whole, or you have
  many categories - beyond 5-6 slices, comparing their sizes by eye
  gets unreliable. A bar chart is usually the safer default.
