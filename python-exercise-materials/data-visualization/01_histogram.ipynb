# Histogram

A histogram groups a single **numeric** variable into bins (intervals)
and shows how many observations fall into each bin. The bars are
usually equal-width and touch each other, since the x-axis is a
continuous range, not separate categories - that's the main thing that
tells a histogram apart from a bar chart.

- **x-axis**: the bin ranges
- **y-axis**: frequency (count) of values in that bin

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
tips.head(3)
```

## Matplotlib

```python
plt.figure(figsize=(8, 6))
plt.hist(tips['total_bill'], bins=10, color='green')   # data, bins, color
plt.title('Total Bill Histogram', size=15)
plt.xlabel('Total Bill', size=10)
plt.ylabel('Frequency', size=10)
plt.grid(True)
plt.show()
```

## Seaborn

`sns.histplot` is the modern replacement for the older `sns.distplot` you may see in older tutorials - `kde=True` overlays a smoothed density curve on top of the bars.

```python
sns.histplot(tips['tip'], kde=True)
plt.title('Tip Histogram', size=20)
plt.show()
```

## Pandas

`Series.hist()` is a thin wrapper around matplotlib.

```python
tips['tip'].hist(color='green', alpha=0.4, bins=15)
plt.title('Tip Histogram', size=20)
plt.show()
```

## Bar chart vs. histogram

| | Bar chart | Histogram |
|---|---|---|
| x-axis | any category (labels) | continuous numeric ranges (bins) |
| bar spacing | gaps between bars | bars touch each other |
| used for | comparing discrete groups | seeing the shape/spread of one numeric variable |
