# Line Plot

A line plot connects ordered data points with straight line segments.
It's the standard choice for **time series** data, where the x-axis is
time and we care about the trend - upward, downward, or seasonal
patterns - rather than comparing discrete categories.

```python
import matplotlib.pyplot as plt
import seaborn as sns

flights = sns.load_dataset("flights")
flights.head(3)
```

## Matplotlib

```python
january = flights[flights['month'] == 'January']

plt.figure(figsize=(8, 6))
plt.plot(january['year'], january['passengers'], 'red')
plt.title('Number of Passengers in January, Yearly', size=15)
plt.xlabel('Year')
plt.ylabel('Number of Passengers')
plt.show()
```

## Seaborn

`hue=` draws one line per category, all on the same axes.

```python
sns.lineplot(data=flights, x='year', y='passengers', hue='month')
plt.title('Number of Passengers per Month, Yearly', size=15)
plt.show()
```

## Pandas

```python
january.plot.line(x='year', y='passengers')
plt.title('Number of Passengers in January, Yearly', size=15)
plt.show()
```

## When to use / avoid

- **Use** it for one or more variables tracked over time (stock
  prices, weather, monthly sales).
- **Avoid** it for comparing unordered categories - a bar chart reads
  better there, since a line implies a continuous path between points
  that don't actually have one.
