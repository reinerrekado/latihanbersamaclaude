# Heatmap

A heatmap encodes a 2D grid of numbers as color instead of text -
darker/more saturated cells usually mean higher values. The most
common use in data analysis is a **correlation heatmap**: every
numeric column plotted against every other numeric column, colored by
how strongly they move together.

```python
import matplotlib.pyplot as plt
import seaborn as sns

tips = sns.load_dataset("tips")
correlation_matrix = tips[['total_bill', 'tip', 'size']].corr()
correlation_matrix
```

## Matplotlib

`pcolor` colors the grid, but doesn't print the numbers on it.

```python
plt.figure(figsize=(6, 5))
plt.pcolor(correlation_matrix, cmap='RdYlGn')
plt.xticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
plt.yticks(range(len(correlation_matrix.columns)), correlation_matrix.columns)
plt.colorbar()
plt.show()
```

## Seaborn

`annot=True` prints the correlation value inside each cell.

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='RdYlGn')
plt.title('Correlation Heatmap', size=15)
plt.show()
```

## Reading it

- Values range from -1 (strong negative correlation) to 1 (strong
  positive correlation); the diagonal is always 1 (a variable is
  perfectly correlated with itself).
- A quick heatmap like this is a common first step in EDA before
  building a model - it flags which variables might be worth
  exploring further, and which ones are redundant with each other.
